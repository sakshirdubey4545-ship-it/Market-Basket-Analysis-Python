```python
url = "https://books.toscrape.com"

response = requests.get(url)

print(response.status_code)

soup = BeautifulSoup(response.text, "html.parser")
```

```python
books = soup.find_all("article", class_="product_pod")

print("Number of books on this page:", len(books))
```

```python
titles = []
prices = []
ratings = []
availability = []
```

```python
for book in books:

    # Book Title
    title = book.h3.a["title"]

    # Price
    price = book.find("p", class_="price_color").text

    # Rating
    rating = book.p["class"][1]

    # Availability
    stock = book.find("p", class_="instock availability").text.strip()

    # Store data in lists
    titles.append(title)
    prices.append(price)
    ratings.append(rating)
    availability.append(stock)
```

```python
print("Titles:", len(titles))
print("Prices:", len(prices))
print("Ratings:", len(ratings))
print("Availability:", len(availability))
```

```python
df = pd.DataFrame({
    "Title": titles,
    "Price": prices,
    "Rating": ratings,
    "Availability": availability
})

df.head()
```

```python
import requests
from bs4 import BeautifulSoup

# Empty lists
titles = []
prices = []
ratings = []
availability = []
categories = []
product_urls = []

base_url = "https://books.toscrape.com/catalogue/page-{}.html"

for page in range(1, 51):

    url = base_url.format(page)
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    books = soup.find_all("article", class_="product_pod")

    for book in books:

        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text
        rating = book.p["class"][1]
        stock = book.find("p", class_="instock availability").text.strip()

        # Product URL
        product_url = "https://books.toscrape.com/catalogue/" + book.h3.a["href"].replace("../", "")

        # Open product page
        product_response = requests.get(product_url)
        product_soup = BeautifulSoup(product_response.text, "html.parser")

        # Category
        category = product_soup.find("ul", class_="breadcrumb").find_all("li")[2].text.strip()

        # Save data
        titles.append(title)
        prices.append(price)
        ratings.append(rating)
        availability.append(stock)
        categories.append(category)
        product_urls.append(product_url)

print("Total Books:", len(titles))
```

```python
df = pd.DataFrame({
    "Title": titles,
    "Price": prices,
    "Rating": ratings,
    "Availability": availability
})

df.head()
```

```python
df = pd.DataFrame({
    "Title": titles,
    "Price": prices,
    "Rating": ratings,
    "Availability": availability,
    "Category": categories,
    "Product_URL": product_urls
})

df.head()
```

```python
df.to_csv("books_data.csv", index=False)
```
